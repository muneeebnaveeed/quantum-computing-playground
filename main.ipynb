# Introduction to Quantum Computing — main notebook

Scratch notebook for the lab. Kernel: the container's `quantum` env (Python 3.11.3, qiskit 2.1.2, qiskit-aer 0.17.2).

```python
import qiskit, qiskit_aer, matplotlib
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

print("qiskit", qiskit.__version__, "| qiskit-aer", qiskit_aer.__version__, "| matplotlib", matplotlib.__version__)
```

## Bell state

```python
qc = QuantumCircuit(2, 2)
qc.h(0)
qc.cx(0, 1)
qc.measure([0, 1], [0, 1])
qc.draw(output="mpl")
```

```python
counts = AerSimulator().run(qc, shots=1000).result().get_counts()
counts
```

```python
from qiskit.visualization import plot_histogram

plot_histogram(counts)
```
